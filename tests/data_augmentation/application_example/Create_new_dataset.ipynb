# Create new Dataset for Example for Data Augmentation in Steel Classification

<br>

```python
import pandas as pd
```

### Create Example Dataset

```python
# Read labeled_OES_data_2.csv
data = pd.read_csv("../labeled_OES_data_2.csv")


element_data_grouped_by_steeltype = data.groupby("steel_class")
dfs_by_steeltype = {}

for type, group_data in element_data_grouped_by_steeltype:
    if type != 3:
        dfs_by_steeltype[type] = group_data.iloc[:, :-1]
```

```python
test_data = pd.DataFrame(columns=dfs_by_steeltype["St Si"].columns)

data = pd.DataFrame(columns=dfs_by_steeltype["St Si"].columns)

for type in dfs_by_steeltype.keys():
        if type == "St 48":
            dfs_by_steeltype[type]["steel_class"] = type
            test_data = pd.concat([dfs_by_steeltype[type].iloc[10:], test_data])
            data = pd.concat([dfs_by_steeltype[type].iloc[:10], data])
        elif type == "St 37":
            dfs_by_steeltype[type]["steel_class"] = type
            test_data = pd.concat([dfs_by_steeltype[type].iloc[20:], test_data])
            data = pd.concat([dfs_by_steeltype[type].iloc[:20], data])
        elif type == "St Si":
            dfs_by_steeltype[type]["steel_class"] = type
            test_data = pd.concat([dfs_by_steeltype[type].iloc[10:], test_data])
            data = pd.concat([dfs_by_steeltype[type].iloc[:10], data])
        elif type == "St 48-TH":
            dfs_by_steeltype[type]["steel_class"] = type
            test_data = pd.concat([dfs_by_steeltype[type].iloc[60:], test_data])
            data = pd.concat([dfs_by_steeltype[type].iloc[:60], data])
        elif type == "Puddelstahl":
            dfs_by_steeltype[type]["steel_class"] = type
            test_data = pd.concat([dfs_by_steeltype[type].iloc[20:], test_data])
            data = pd.concat([dfs_by_steeltype[type].iloc[:20], data])

test_data.tail()

#cut_data = int(dfs_by_steeltype[type].shape[0]/6)
 #       dfs_by_steeltype[type]["steel_class"] = type
  #      test_data = pd.concat([dfs_by_steeltype[type].iloc[cut_data:], test_data])
   #     data = pd.concat([dfs_by_steeltype[type].iloc[:cut_data], data])
```

```python
data.to_csv("example_dataset.csv", index=False)
test_data.to_csv("example_testing_dataset.csv", index=False)
```
